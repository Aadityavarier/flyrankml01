## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The Decision: Which specific web pages should the SEO/content team prioritize for an intensive rewrite or refresh this week?

The Action: A content writer reviews the selected page, updates out-of-date information, improves keyword alignment, and republishes it.

The Unit of Analysis: A unique URL (page) per client.

The Cost of a Wrong Call (False Positive / False Negative):

False Positive: Recommending a stable/growing page for a refresh wastes valuable hours of a content writer's time fixing something that wasn't broken.

False Negative: Missing a page that is genuinely tanking results in a permanent loss of search impressions, organic traffic, and potential business revenue.

Why ML helps: A simple fixed rule cannot balance multi-dimensional features like long-term historical impressions, keyword volume, and sudden position drops across thousands of pages simultaneously. ML can learn the complex interaction patterns that signal imminent decay.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import sys
import subprocess
import pandas as pd

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"


if "google.colab" in sys.modules and not os.path.isdir(REPO_DIR):
    print("Cloning the internship starter repository...")
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)


target_file = None
for root, dirs, files in os.walk("."):
    if "content_refresh_anonymized.csv" in files:
        target_file = os.path.join(root, "content_refresh_anonymized.csv")
        break

if target_file:
    print(f"Success! Found dataset at: {target_file}\n" + "="*50)
    df = pd.read_csv(target_file)

    trend_counts = df["trend_direction"].value_counts()
    total_pages = len(df)
    down_pages = trend_counts.get("down", 0)
    pct_down = (down_pages / total_pages) * 100

    print(f"Total pages in sample: {total_pages:,}")
    print(f"Number of decaying ('down') pages: {down_pages:,} ({pct_down:.1f}%)")


    total_impressions = df["impressions_90d"].sum()
    down_impressions = df[df["trend_direction"] == "down"]["impressions_90d"].sum()
    pct_impressions_at_risk = (down_impressions / total_impressions) * 100

    print(f"Total impressions across all pages: {total_impressions:,}")
    print(f"Impressions at risk (belonging to 'down' pages): {down_impressions:,} ({pct_impressions_at_risk:.1f}%)")
else:
    print("Error: Could not locate 'content_refresh_anonymized.csv'.")
```

The data shows that a significant portion of pages are actively decaying, representing a substantial percentage of the overall traffic footprint. This proves there is a massive operational lift in automating the discovery of these specific pages.
